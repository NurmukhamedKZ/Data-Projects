```python
import camelot
import pandas as pd
import matplotlib.pyplot as plt

grant_B057 = camelot.read_pdf("data/B057_all.pdf", pages="all", flavor="lattice", suppress_stdout=False)
```

```python
# for rollback the table
collect_grant_B057 = pd.DataFrame({0: [],
                              1: [],
                              2: [],
                              3: [],
                              4: []})

for page in grant_B057:
    collect_grant_B057 = pd.concat([collect_grant_B057, page.df], ignore_index=True)
```

```python
# Parsing
collect_grant_B057.rename(columns={0:"#",1:"id",2:"name",3:"score",4:"university"}, inplace=True) 

collect_grant_B057["name"] = collect_grant_B057["name"].str.replace("\n","")
collect_grant_B057["#"] = collect_grant_B057["#"].str.replace("\n","")


for i, row in collect_grant_B057.iterrows():
    if row["#"] == "":
        collect_grant_B057.loc[i-1, "name"] += " " + collect_grant_B057.loc[i,"name"]
        collect_grant_B057.drop(i, inplace=True)
    elif len(row["#"]) >=7:
        collect_grant_B057.drop(i, inplace=True) 


        
# Sharing
collect_grant_B057.to_csv("data/grants_B057.csv", index=False)
```

```python

parsed_grants = pd.read_csv("data/grants_B057.csv")

x = parsed_grants["score"].drop_duplicates()
y = parsed_grants["score"].value_counts(sort=False)/sum(parsed_grants["score"].value_counts(sort=False))*100

plt.figure(figsize=(12, 6))
plt.bar(x,y , color='skyblue')
plt.title('B057 IT')
plt.xlabel('Score [0-140]')
plt.ylabel('Percent of scores %')
plt.xticks()
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
