```python
import camelot
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
grants_2 = camelot.read_pdf("data/all_grants.pdf", pages="all", flavor="stream", suppress_stdout=False)
```

```python
grants_test = pd.DataFrame({0: [],
                              1: [],
                              2: [],
                              3: [],
                              4: []})

for page in grants_2:
    grants_test = pd.concat([grants_test, page.df], ignore_index=True)
```

```python
# Parsing
grants_test.rename(columns={0:"#",1:"id",2:"name",3:"score",4:"university"}, inplace=True) 

grants_test["name"] = grants_test["name"].str.replace("\n","")
grants_test["#"] = grants_test["#"].str.replace("\n","")

```

```python
grants_test.tail(20)
```

```python
# cnt_of_starts = 0
pages = []
for i in range(1,1749):
    pages.append(str(i))

for i, row in grants_test.iterrows():
    if row["name"] in ("ЖАЛПЫ КОНКУРС", "АУЫЛ КВОТАСЫ"):
        grants_test.drop(i, inplace=True) 
    elif row["id"] == "ТЖК":
        grants_test.drop(i, inplace=True)
    elif row["name"] in pages:
        grants_test.drop(i, inplace=True)
    
```

```python
grants_test = pd.read_csv("data/all_grants_raw.csv")
```

```python
def check_exist(index):
    try:
        if grants_test.loc[index, "name"] == grants_test.loc[index, "name"]:
            return True
    except KeyError:
        return False

for i, row in grants_test.iterrows():
    if row["name"].count(" ") == 0 and len(row["name"]) != 0:
        if check_exist(i-1):
            if len(grants_test.loc[i-1, "name"]) == 0:    
                grants_test.loc[i-1, "name"] = grants_test.loc[i-2, "name"] + " " + grants_test.loc[i, "name"] 
                grants_test.drop([i-2, i], inplace=True)
            elif grants_test.loc[i-1, "name"].count(" ") in (0,1) and grants_test.loc[i+2, "name"].count(" ") != 0:
                grants_test.loc[i-1, "name"] += " " + grants_test.loc[i, "name"] 
                grants_test.drop(i, inplace=True)
        elif check_exist(i-2):
            if len(grants_test.loc[i-2, "name"]) == 0:    
                grants_test.loc[i-2, "name"] = grants_test.loc[i-3, "name"] + " " + grants_test.loc[i, "name"] 
                grants_test.drop([i-3, i], inplace=True)
            elif grants_test.loc[i-2, "name"].count(" ") in (0,1) and grants_test.loc[i+2, "name"].count(" ") != 0:
                grants_test.loc[i-2, "name"] += " " + grants_test.loc[i, "name"] 
                grants_test.drop(i, inplace=True)
```
