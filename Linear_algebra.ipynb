```python
import numpy as np

def cross_product(a,b):
    return np.array([a[1]*b[2]-a[2]*b[1],
                     a[2]*b[0]-a[0]*b[2],
                     a[0]*b[1]-a[1]*b[0]])

vec1 = np.array([1,-7,1])
vec2 = np.array([5,2,4])

print(np.cross(vec1,vec2))
print(cross_product(vec1,vec2))
```
