```python
import os
import matplotlib.pyplot as plt
import tensorflow as tf
import numpy as np
```

```python
paths = []

for dirname, _, filenames in os.walk('Apparel_small/'):
    print(dirname, _, filenames)
    for file_name in filenames:
        paths.append(os.path.join(dirname,file_name))

paths
```

```python
idx = 0
print(paths[idx])
img = plt.imread(paths[idx])
print(img.shape)
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
import random
random.seed(9)
np.random.seed(9)
tf.random.set_seed(9)

np.random.shuffle(paths)
paths
```

```python
labels = []

for path in paths:
    labels.append(path.split("/")[-2].split("_"))

labels
```

```python
test_size = 0.2
thresh = round(len(paths)*test_size)

test_paths = paths[:thresh]
train_paths = paths[thresh:]

print(len(train_paths), len(test_paths))
train_paths[:10]
```

```python
test_labels = labels[:thresh]
train_labels = labels[thresh:]

train_labels[:10]
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()

train_classes = mlb.fit_transform(train_labels)
test_classes = mlb.transform(test_labels)

train_classes[:10]
```

```python
from keras.utils import Sequence


class DataGenerator(Sequence):
    def __init__(self, paths, labels, batch_size=32, pixel_size=64):
        self.batch_size = batch_size
        self.paths = paths
        self.labels = labels
        self.pixel_size = pixel_size

    def __len__(self):
        return int(np.round(len(self.paths) / self.batch_size))

    def __getitem__(self, index):
        X = self.paths[index * self.batch_size : (index+1) * self.batch_size]
        y = self.labels[index * self.batch_size : (index+1) * self.batch_size]
        X = np.array([self.load_img(i, self.pixel_size) for i in X])
        return X, y

    @staticmethod
    def load_img(path, pixel_size):
        img = plt.imread(path)[np.newaxis, ...]
        img_resized = tf.image.resize(img, (pixel_size, pixel_size), method='nearest')[0]
        img_resized = np.array(img_resized, dtype='float')
        img_resized /= 255.0
        return img_resized
```

```python
train_datagen = DataGenerator(train_paths, train_classes)
test_datagen = DataGenerator(test_paths, test_classes)
```

```python
for X,y in train_datagen:
    print(X.shape)
    print(y.shape)
    print(y[:10])
    break
```

```python
from keras.layers import Flatten, Dense, Input, Softmax
from keras.models import Model
from keras.optimizers import Adam
from keras.losses import CategoricalCrossentropy

input = Input(shape=(64,64,3))
flatten = Flatten()(input)
dense1 = Dense(32, activation="relu")(flatten)
dense2 = Dense(6, activation="linear")(dense1)
output = Softmax()(dense2)

model = Model(inputs=input, outputs=output)

model.compile(optimizer=Adam(),loss=CategoricalCrossentropy(from_logits=False), metrics=["accuracy"])

model.summary()
```

```python
history = model.fit(train_datagen, epochs=5, validation_data=test_datagen)
```

```python
plt.plot(range(1,len(history.history["loss"])+1), history.history["loss"], color="red", label="train")
plt.plot(range(1,len(history.history["loss"])+1), history.history["val_loss"], color="blue", label="test")
plt.ylabel("loss")
plt.legend()
plt.show()
```

```python
import seaborn as sns

img = np.array([
    [8, 12, 8, 8, 8],
    [8, 12, 8, 8, 8],
    [8, 13, 8, 8, 8],
    [8, 12, 8, 8, 8],
    [8, 13, 8, 8, 8],
])

fig, axs = plt.subplots(1, 2, figsize=(10, 5))
sns.heatmap(img, cmap='gray', annot=True, cbar=False, ax=axs[0])


kernel = np.array([
   [-1, 0, 1],
   [-2, 0, 2],
   [-1, 0, 1]
])

sns.heatmap(kernel, cmap='gray', annot=True, cbar=False, linewidths=1, linecolor='k', ax=axs[1]);
```

```python
def apply_kernel(img1, kernel1):
    size_y = img1.shape[0]
    size_x = img1.shape[1]
    feature_map = np.copy(img1)
    for x in range(1,size_x-1):
        for y in range(1,size_y-1):
            curr_img = img1[x-1:x+2, y-1:y+2]
            convolution = np.sum(curr_img * kernel1)

            if convolution < 0:
                convolution = 0
            elif convolution > 255:
                convolution = 255
            feature_map[x,y] = convolution
        
    return feature_map

img_transformed = apply_kernel(img, kernel)

sns.heatmap(img_transformed, cmap='gray', annot=True, cbar=False, linewidths=1, linecolor='k');
```

```python
def plot_image(img, img_transformed):

    fig, axs = plt.subplots(2, 3, figsize=(15, 7))
    axs[0][0].imshow(img, cmap='gray')
    axs[0][1].imshow(img_transformed, cmap='gray')
    axs[0][2].imshow(np.clip(img_transformed, 0, 1), cmap="gray")

    ymin,ymax,xmin,xmax = 100,150,100,150

    axs[1][0].imshow(img[ymin:ymax,xmin:xmax], cmap='gray')
    axs[1][1].imshow(img_transformed[ymin:ymax,xmin:xmax], cmap='gray')
    axs[1][2].imshow(np.clip(img_transformed[ymin:ymax,xmin:xmax], 0, 1), cmap="gray")

    
    plt.show()

plot_image(img, feature_map)
```

### To make more detailed image

```python
kernel = np.array([[-1,-1,-1],
                   [-1,9,-1],
                   [-1,-1,-1]])

feature_map = apply_kernel(img, kernel)
plot_image(img, feature_map)
```

```python


fig, ax = plt.subplots(2, 3, figsize=(15, 7))
ax = np.ravel(ax)

for i, ks in enumerate(np.arange(3, 20, 3)):
    random.seed(1)
    tf.random.set_seed(1)
    model = Sequential([
        Conv2D(filters=1,
            kernel_size=(ks, 3),
            use_bias=False,
            input_shape=(img.shape[0], img.shape[1], 3))
    ])
    amount_w = model.layers[0].get_weights()[0].size

    pred = model.predict(img[None], verbose=0)[0]
    ax[i].set_title(f'weights {amount_w}\nkernel ({ks}, 3). img size {pred.shape}')
    ax[i].imshow(pred, cmap='gray')
```

```python
fig, ax = plt.subplots(2, 2, figsize=(15, 9))
ax = np.ravel(ax)

for i, s in enumerate(np.arange(1, 5)):
    random.seed(1)
    tf.random.set_seed(1)
    model = Sequential([
        Conv2D(filters=1,
            kernel_size=(3, 3),
            use_bias=False,
            padding='same',
            strides=(s, s),
            input_shape=(img.shape[0], img.shape[1], 3))
    ])
    amount_w = model.layers[0].get_weights()[0].size

    pred = model.predict(img[None], verbose=0)[0]
    ax[i].set_title(f'weights {amount_w}\nstrides ({s}, {s}). img size {pred.shape}')
    ax[i].imshow(pred, cmap='gray')
```

```python
fig, ax = plt.subplots(2, 3, figsize=(15, 7))
ax = np.ravel(ax)

for i, d in enumerate(np.arange(1, 18, 3)):
    random.seed(1)
    tf.random.set_seed(1)
    model = Sequential([
        Conv2D(filters=1,
            kernel_size=(3, 3),
            use_bias=False,
            padding='same',
            dilation_rate=(d, d),
            input_shape=(img.shape[0], img.shape[1], 3))
    ])
    amount_w = model.layers[0].get_weights()[0].size

    pred = model.predict(img[None], verbose=0)[0]
    ax[i].set_title(f'weights {amount_w}\ndilitaion ({d}, {d}). img size {pred.shape}')
    ax[i].imshow(pred, cmap='gray')
```
