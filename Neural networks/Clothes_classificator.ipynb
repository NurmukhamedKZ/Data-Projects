```python
from keras.datasets import fashion_mnist

(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

class_names = ['T-shirt/top',
               'Trouser',
               'Pullover',
               'Dress',
               'Coat',
               'Sandal',
               'Shirt',
               'Sneaker',
               'Bag',
               'Ankle boot']

X_train
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(1,5, figsize=(10,10))

for i in range(5):
    ax[i].imshow(X_train[i], cmap="Greys")
    ax[i].axis("off")
```

```python
y_train[:5]
```

```python
from keras.utils import to_categorical
X_train = X_train / 255
X_test = X_test / 255

y_train_cat = to_categorical(y_train)
y_test_cat = to_categorical(y_test)

y_train_cat
```

```python
from keras.models import Sequential
from keras.layers import Dense, Flatten

model = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10, activation="sigmoid")
])

model.compile(optimizer="sgd", loss="categorical_crossentropy", metrics=["accuracy"])

model.summary()
```

```python
model.fit(X_train, y_train_cat, epochs=15)
```

```python
pred_prob = model.predict(X_train)
pred_class = pred_prob.argmax(axis=1)
pred_class
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y_train, pred_class))
```

```python
from keras.optimizers import SGD
model = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation="sigmoid"),
    Dense(64, activation="sigmoid"),
    Dense(32, activation="sigmoid"),
    Dense(16, activation="sigmoid"),
    Dense(10, activation="sigmoid")
])

optimizer = SGD(learning_rate=0.01)

model.compile(optimizer=optimizer, loss="categorical_crossentropy", metrics=["accuracy"])

model.summary()
```

```python
# model.fit(X_train, y_train_cat, epochs=30)
```

```python
pred_prob = model.predict(X_test)
pred_class = pred_prob.argmax(axis=1)
pred_class
```

```python
print(f"accuracy: {accuracy_score(y_test, pred_class)*100}% | correctly guessed: {(y_test == pred_class).sum()} out of {y_test.shape[0]}")
```

```python
import tensorflow as tf

batch_size = 64

train_dataset = tf.data.Dataset.from_tensor_slices((X_train,y_train_cat))
train_dataset = train_dataset.shuffle(buffer_size=X_train.shape[0]).batch(batch_size=batch_size)
train_dataset
```

```python
# fig, ax =  plt.subplots(1,5, figsize=(10,5))

for X,y in train_dataset:
    # for i in range(5):
    #     ax[i].imshow(X[i], cmap="Greys")
    #     ax[i].axis("off")
    print(X)
    break
```

```python
from keras.losses import CategoricalCrossentropy
from keras.optimizers import SGD

optimizer = SGD(learning_rate=0.01)
loss_fn = CategoricalCrossentropy()
```

```python
grad_log = []

epochs = 2

for epoch in range(epochs):
    for step, (X_batch_train,y_batch_train) in enumerate(train_dataset):
        with tf.GradientTape() as tape:
            preds = model(X_batch_train)
            loss = loss_fn(y_batch_train, preds)
            grads = tape.gradient(loss, model.trainable_weights)
          
            g_g = []

            for g_s in grads:
                if len(g_s.numpy().shape) == 2:
                    g_g.append(g_s.numpy()[0, 0]) 

        grad_log.append(g_g)

        optimizer.apply_gradients(zip(grads, model.trainable_weights))

        if step % 200 == 0:
            print(f'Epoch {epoch + 1}/{epochs}', end='. ')
            print(f'Step {step}. Train loss: {loss}') 
```

```python
pred_prob = model.predict(X_test)
pred_class = pred_prob.argmax(axis=1)
print(f"accuracy: {accuracy_score(y_test, pred_class)*100}% | correctly guessed: {(y_test == pred_class).sum()} out of {y_test.shape[0]}")
```

```python
grad_log = np.array(grad_log)
grad_log
```

```python
plt.figure(figsize=(16, 5))
plt.title('Grad by layer')
plt.xlabel('№ layer')
plt.ylabel('grad')
plt.grid()

plt.plot(np.abs(grad_log[0, :]), label='step 0')
plt.plot(np.abs(grad_log[10, :]), label='step 10')
plt.plot(np.abs(grad_log[100, :]), label='step 100')
plt.plot(np.abs(grad_log[700, :]), label='step 700')
plt.legend()
plt.show()
```
