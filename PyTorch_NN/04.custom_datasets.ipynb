```python
def decorator_func(function):
    def some_func(function):
        print(f"the name of the original function: {function.__name__}")
        return function()
    return some_func

@decorator_func
def display(n):
    print(f"n: {n}")

display("hello")
```

```python
def decorator_func(function):
    def wrapper(*args, **kwargs):
        print(f"wrapper: {args} , {kwargs}")
        return function(args, kwargs)
    return wrapper


@decorator_func
def display2(name,age):
    print(f"original: {name} , {age}")


display2("hello", 124, privet=666)
```
