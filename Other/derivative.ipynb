```python

def derivative_x_power(num,power):
    x = num
    y = num ** power
    m = power*num**(power - 1)
    b = y - x * m
    return f"y = {m}x +{b}" if b > 0 else f"y = {m}x {b}"

derivative_x_power(-5,2)
```

```python
def estimate_derivative(table):
    return (table["y"][1]-table["y"][0])/(table["x"][1] - table["x"][0])

table1 = {"x": [-8,-4],
          "y": [85,77]}

estimate_derivative(table1) 
```
